```python
import json
import os

INPUT_FILEPATH = "/content/drive/MyDrive/Datasets/GenEval/evaluation_metadata.jsonl"
# The output file where the 100 Color Attribution prompts will be saved.
OUTPUT_FILEPATH = "/content/drive/MyDrive/Datasets/GenEval/geneval_color_attr_prompts.txt"

TARGET_TAG = "color_attr"
```

```python
def extract_geneval_prompts(input_path: str, output_path: str, target_tag: str):
    """
    Reads the GenEval metadata file, filters for a specific compositional tag,
    and saves the extracted text prompts to a new file.
    """
    if not os.path.exists(input_path):
        print(f"Error: Input file not found at {input_path}")
        print("Please ensure you have saved 'evaluation_metadata.jsonl' in the correct location.")
        return

    extracted_prompts = []

    with open(input_path, 'r') as f:
        for line in f:
            try:
                data = json.loads(line.strip())
            except json.JSONDecodeError:
                print(f"Warning: Skipping invalid JSON line: {line.strip()}")
                continue

            if data.get("tag") == target_tag:
                prompt = data.get("prompt")
                if prompt:
                    extracted_prompts.append(prompt)

    # Check the count to verify we extracted the correct subset
    num_prompts = len(extracted_prompts)

    with open(output_path, 'w') as f:
        f.write('\n'.join(extracted_prompts))

    print("\n--- Prompt Extraction Complete ---")
    print(f"Target Tag: '{target_tag}'")
    print(f"Total prompts extracted: {num_prompts}")

    # Display a few examples for verification
    print("\n--- Example Prompts ---")
    for i, p in enumerate(extracted_prompts[:5]):
        print(f"  {i+1}. {p}")
    print("...")

    return extracted_prompts
```

```python
extracted_prompts = extract_geneval_prompts(INPUT_FILEPATH, OUTPUT_FILEPATH, TARGET_TAG)
```
